```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
train = pd.read_csv("/kaggle/input/competitions/data-science-london-scikit-learn/train.csv",header=None)
test = pd.read_csv("/kaggle/input/competitions/data-science-london-scikit-learn/test.csv",header=None)
labels = pd.read_csv("/kaggle/input/competitions/data-science-london-scikit-learn/trainLabels.csv",header=None)
```

```python
train.head()
```

```python
test.head()
```

```python
labels.head()
```

```python
pipeline = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5, weights = "uniform", p=2, metric='minkowski')
)
```

```python
pipeline.fit(train,labels)
```

```python
preds = pipeline.predict(test)
```

```python
submission = pd.DataFrame({
    "Solution":preds
})

submission.index = submission.index + 1

submission =  submission.reset_index().rename(columns={"index": "Id"})

submission
```

```python
submission.to_csv("submission.csv",index=False)
```
